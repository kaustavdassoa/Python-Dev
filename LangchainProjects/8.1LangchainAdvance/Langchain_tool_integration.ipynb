```python
from langchain.tools import tool

@tool
def get_weather(location) -> str:
    """Get the weather for a location."""
    return f"The weather in {location} is Holiday Mood."
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3-32b")
```

```python
model_with_tools=model.bind_tools([get_weather])
```

```python
#Step 1 : Create user messages
messages = [{"role": "user", "content": "What is the weather of Bangalore?"}]

#Step 2 : Get AI response - which tool needs to be called
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

# Debugging steps 
# for tool_info in ai_msg.tool_calls:
#     print(f"Tools called : '{tool_info['name']}' arguments passed : {tool_info['args']}")


#Step 3 : Get AI response - by calling the tool
for tool_call in ai_msg.tool_calls:
    tool_results = get_weather.invoke(tool_call)
    messages.append(tool_results)


#Step 4 : Get final AI response
final_response = model_with_tools.invoke(messages)

print("Model Response : ", final_response.text)

print("\nReasoning : ")
print("-------------------------------------------------------")
print(final_response.additional_kwargs['reasoning_content'])
```
