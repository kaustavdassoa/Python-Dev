```python
from langchain.agents import create_agent
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")


```

```python

agent=create_agent(
    model= "gpt-4o",
    tools= [],
    system_prompt= "You are a helpful assistant."
)

#agent.invoke("What is the weather of France?")
response = agent.invoke({"messages": [("user", "What is the capital of India?")]})

print(response["messages"][-1].content)
```

```python

def get_weather(location) -> str:
    """Get the weather for a location."""
    return f"The weather in {location} is Holiday mood."


newAgent = create_agent( model="gpt-4o", 
                         tools=[get_weather], 
                         system_prompt="You are a helpful assistant."
                        )
                        

response=newAgent.invoke({"messages": [("user", "What is the weather of Bangalore?")]})

print(response["messages"][-1].content)
```
