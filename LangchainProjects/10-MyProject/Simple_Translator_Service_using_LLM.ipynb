# Translator Service 

Simple translator service LangChain Expression Language :

1. Call open-source llama model 
2. Leverage Grok Interface - free tier developer API Interface
3. Leverage prompt template 
4. Leverage parser & Langchain 


```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
from langserve import add_routes
import os
from dotenv import load_dotenv
from fastapi import FastAPI
```

```python
# Connecting to Gemma2-9b-It model

modelName = "llama-3.3-70b-versatile"
groq_api_key=os.getenv("GROQ_API_KEY")
model=ChatGroq(model=modelName,groq_api_key=groq_api_key)
```

```python
# Ctrl  + Enter --> new cell above
# Shift + Enter --> new cell below

system_template = "You are a helpful translator service, translate the following into : {language} "

prompt_template = ChatPromptTemplate.from_messages([
    ('system', system_template),
    ('user', '{text}')
])
```

```python
parser=StrOutputParser()
```

```python
##create chain
chain=prompt_template|model|parser
```

```python
language ="Assamese"
translation_text="What a beautiful day today?"

response=chain.invoke({"language" : language, "text" : translation_text})
print(f"translation of  '{translation_text}' in {language} is \n{response}")
```
